```python
# Create a simple ATM menu.

balance = 5000

while True:
    print("\n1. Withdraw")
    print("2. Deposit")
    print("3. Balance")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        amount = int(input("Enter withdrawal amount: "))

        if amount > 0 and amount <= balance:
            balance -= amount
            print("Withdrawal successful")
        else:
            print("Invalid amount or insufficient balance")

    elif choice == 2:
        amount = int(input("Enter deposit amount: "))

        if amount > 0:
            balance += amount
            print("Deposit successful")
        else:
            print("Invalid amount")

    elif choice == 3:
        print("Current Balance:", balance)

    elif choice == 4:
        print("Thank you!")
        break

    else:
        print("Invalid choice")
```

```python
# Print all prime numbers between 1 and n.

n = int(input("Enter a number: "))

for num in range(2, n + 1):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

```python
# Check whether a number is an Armstrong number.

n = int(input("Enter a number: "))

if n < 0:
    print("Not an Armstrong number")

else:
    temp = n
    power = len(str(n))
    total = 0

    while temp > 0:
        digit = temp % 10
        total += digit ** power
        temp //= 10

    if total == n:
        print("Armstrong Number")
    else:
        print("Not Armstrong Number")
```

```python
# Reverse a number and check if it is a palindrome.

n = int(input("Enter a number: "))

temp = n
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp //= 10

print("Reverse:", reverse)

if n >= 0 and n == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Find the Greatest Common Divisor (GCD) of two numbers.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

a = abs(a)
b = abs(b)

while b != 0:
    a, b = b, a % b

print("GCD:", a)
```

```python
# Write a function to check if a string is a palindrome.

def palindrome(s):
    return s == s[::-1]

print(palindrome("madam"))
print(palindrome("hello"))
```

```python
# Write a function to count vowels and consonants in a string.

def count_letters(s):
    vowels = 0
    consonants = 0

    for i in s.lower():
        if i in "aeiou":
            vowels += 1
        elif i.isalpha():
            consonants += 1

    return vowels, consonants

v, c = count_letters("Python Programming")

print("Vowels:", v)
print("Consonants:", c)
```

```python
# Create a function to calculate simple and compound interest.

def interest(p, r, t):
    simple = (p * r * t) / 100
    compound = p * ((1 + r/100) ** t) - p

    return simple, compound

si, ci = interest(10000, 5, 2)

print("Simple Interest:", si)
print("Compound Interest:", round(ci, 2))
```

```python
# Write a function to return all factors of a number.

def factors(n):
    result = []

    if n == 0:
        return result

    for i in range(1, abs(n) + 1):
        if n % i == 0:
            result.append(i)

    return result

print(factors(12))
```

```python
# Find the second-largest and second-smallest elements.

numbers = [10, 20, 30, 40, 50]

unique = []

for i in numbers:
    if i not in unique:
        unique.append(i)

if len(unique) < 2:
    print("Not enough distinct elements")

else:
    smallest = float('inf')
    second_smallest = float('inf')

    largest = float('-inf')
    second_largest = float('-inf')

    for i in unique:
        if i < smallest:
            second_smallest = smallest
            smallest = i
        elif i < second_smallest:
            second_smallest = i

        if i > largest:
            second_largest = largest
            largest = i
        elif i > second_largest:
            second_largest = i

    print("Second Smallest:", second_smallest)
    print("Second Largest:", second_largest)
```

```python
# Copy only the even-numbered lines from one file to another.

with open("sample.txt", "r") as file:
    lines = file.readlines()

with open("even_lines.txt", "w") as file:
    for i in range(len(lines)):
        if (i + 1) % 2 == 0:
            file.write(lines[i])

print("Even numbered lines copied successfully")
```

```python
# Create a random password generator using random and string modules.

import random
import string

length = int(input("Enter password length: "))

if length < 1:
    print("Invalid password length")

else:
    characters = string.ascii_letters + string.digits + string.punctuation

    password = ""

    for i in range(length):
        password += random.choice(characters)

    print("Generated Password:", password)
```

```python
# Use the datetime module to calculate the number of days between two dates.

from datetime import datetime

date1 = input("Enter first date (DD-MM-YYYY): ")
date2 = input("Enter second date (DD-MM-YYYY): ")

d1 = datetime.strptime(date1, "%d-%m-%Y")
d2 = datetime.strptime(date2, "%d-%m-%Y")

difference = abs((d2 - d1).days)

print("Number of days:", difference)
```
