```python
#Write a program to check whether a given number is prime.
n = int(input("Enter a number: "))

if n <= 1:
    print("Not Prime")
else:
    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

```python
#Write a program to find the factorial of a number using a loop.
n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial:", fact)
```

```python
#Write a program to check whether a string is a palindrome.
string = input("Enter String:")

reverse = string[::-1]

if (string == reverse):
    print("String is palindrom")
else:
    print("String is not palindrom")
```

```python
#Write a program to count the number of vowels and consonants in a string.
s = input("Enter a string: ")

vowels = 0
consonants = 0

for i in s.lower():
    if i in "aeiou":
        vowels += 1
    elif i.isalpha():
        consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
#Write a program to find the largest and smallest element in a list without using max() or min().
numbers = [10, 25, 5, 40, 15]

largest = numbers[0]
smallest = numbers[0]

for i in numbers:
    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

print("Largest:", largest)
print("Smallest:", smallest)
```

```python
#Write a program to count how many times each element appears in a list using a dictionary.
numbers = [1, 2, 2, 3, 3, 3, 4]

frequency = {}

for i in numbers:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1

print(frequency)
```

```python
#Write a program to find the second-largest number in a list.
numbers = [10, 20, 40, 30, 50]

largest = float('-inf')
second = float('-inf')

for i in numbers:
    if i > largest:
        second = largest
        largest = i

    elif largest > i > second:
        second = i

print("Second Largest:", second)
```

```python
#Write a program to sort a list without using sort() or sorted().
numbers = [5, 2, 8, 1, 3]

n = len(numbers)

for i in range(n):
    for j in range(0, n-i-1):
        if numbers[j] > numbers[j+1]:
            numbers[j], numbers[j+1] = numbers[j+1], numbers[j]

print("Sorted List:", numbers)
```

```python
#Write a function that accepts a list of numbers and returns a list containing only the even numbers.
def even_numbers(numbers):
    even = []

    for i in numbers:
        if i % 2 == 0:
            even.append(i)

    return even

print(even_numbers([1, 2, 3, 4, 5, 6]))
```

```python
#Write a program to find all duplicate values in a list.
numbers = [1, 2, 3, 2, 4, 1, 5, 3]

duplicates = []

for i in numbers:
    if numbers.count(i) > 1 and i not in duplicates:
        duplicates.append(i)

print("Duplicates:", duplicates)
```

```python
#Write a program that takes a sentence and finds the longest word.
sentence = input("Enter a sentence: ")

words = sentence.split()

longest = ""

for i in words:
    if len(i) > len(longest):
        longest = i

print("Longest word:", longest)
```

```python
'''Create a simple student marks program that:
accepts marks for 5 subjects,
calculates the total and percentage,
assigns a grade,and displays whether the student passed or failed.
'''

marks = []

for i in range(1, 6):
    m = int(input(f"Enter marks for subject {i}: "))
    marks.append(m)

total = sum(marks)
percentage = total / 5

print("Total Marks:", total)
print("Percentage:", percentage)

if percentage >= 90:
    grade = "A"
elif percentage >= 80:
    grade = "B"
elif percentage >= 70:
    grade = "C"
elif percentage >= 60:
    grade = "D"
else:
    grade = "F"

if all(m >= 35 for m in marks):
    result = "Pass"
else:
    result = "Fail"

print("Grade:", grade)
print("Result:", result)
```
